# Intervals And Circle Geometry

This notebook demonstrates the basic set-theoretic primitives of the `geo` package on the real line and on the unit circle.

```python
import math

from geo import FloatCircleInterval, FloatCircleSet, FloatInterval, FloatSet
```

```python
left = FloatSet.from_single_interval(0.0, 3.0)
right = FloatSet.from_single_interval(2.0, 5.0)

print('left:', left)
print('right:', right)
print('union:', left | right)
print('intersection:', left & right)
print('difference:', left - right)
print('symmetric difference:', left ^ right)
```

```python
arc_a = FloatCircleInterval(0.0, math.pi / 2.0)
arc_b = FloatCircleInterval(math.pi / 3.0, math.pi)

print('arc_a:', arc_a)
print('arc_b:', arc_b)
print('arc union:', FloatCircleSet(arc_a).union(arc_b))
print('arc intersection:', FloatCircleSet(arc_a).intersection(arc_b))
print('arc difference:', FloatCircleSet(arc_a).difference(arc_b))
```

```python
wrapped = FloatCircleInterval(3.0 * math.pi / 2.0, math.pi / 2.0)
print('wrapped interval tuple:', wrapped.to_tuple())
print('stored linear pieces:', list(wrapped))
print('contains 0:', 0.0 in wrapped)
print('contains pi:', math.pi in wrapped)
```
