# Modern Space, Object, Mesh, Export, And Notebook Helpers

This notebook demonstrates the current end-to-end workflow: default space -> native object -> mesh -> plotting/export.


```python
from pathlib import Path
import sys
import math

repo_root = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / 'pyproject.toml').exists()
)
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from geo import (
    Ball,
    FloatPoint,
    SphereSpace,
    TorusSpace,
    cap,
    disk,
    plot,
    point,
    use_space,
    write_obj,
)
```

```python
plane_disk = disk((0.0, 0.0), 1.0)
origin = point(0.0, 0.0)

sphere = use_space(SphereSpace())
north = sphere.point_from_angles(0.0, math.pi / 2.0)
hemisphere = cap(north, math.pi / 2.0)

print('default plane disk contains origin:', FloatPoint(0.0, 0.0) in plane_disk)
print('sphere cap contains north:', north in hemisphere)
print('origin object contains origin:', FloatPoint(0.0, 0.0) in origin)
```

```python
figure, axis = plot(plane_disk, resolution=12)
print('interactive plot figure:', type(figure).__name__)
```

```python
sphere = SphereSpace()
north = sphere.point_from_angles(0.0, math.pi / 2.0)
cap = sphere.cap(north, math.pi / 3.0)
cap_mesh = cap.mesh(resolution=12)

torus = TorusSpace()
patch = torus.patch((0.0, math.pi / 2.0), (0.0, math.pi / 2.0))
patch_mesh = patch.mesh(resolution=12)

print('cap samples:', len(cap.sample_points(resolution=12)))
print('cap mesh:', len(cap_mesh.vertices), len(cap_mesh.cells))
print('patch samples:', len(patch.sample_points(resolution=12)))
print('patch mesh:', len(patch_mesh.vertices), len(patch_mesh.cells))
```

```python
ball_mesh = Ball(FloatPoint(0.0, 0.0), 1.0).mesh(resolution=12)
wireframe = ball_mesh.wireframe_data()
plotly_data = ball_mesh.plotly_data(name='disk')
threejs_data = ball_mesh.threejs_data()

print('wireframe edges:', len(wireframe['edges']))
print('plotly traces:', len(plotly_data))
print('threejs positions:', len(threejs_data['position']))
```

```python
figure, axis = ball_mesh.plot_matplotlib()
output_dir = repo_root / 'examples' / '_generated'
output_dir.mkdir(exist_ok=True)
write_obj(ball_mesh, output_dir / 'disk_mesh_from_notebook.obj')
print('figure type:', type(figure).__name__)
print('obj exists:', (output_dir / 'disk_mesh_from_notebook.obj').exists())
```
